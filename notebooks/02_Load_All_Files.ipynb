```python
def clean_fbref_columns(df):
    df = df.copy()

    # Flatten MultiIndex columns
    if isinstance(df.columns, pd.MultiIndex):
        new_columns = []

        for col in df.columns:
            parts = []

            for part in col:
                part = str(part).strip()

                # Ignore empty / unnamed levels
                if part and not part.startswith("Unnamed"):
                    parts.append(part)

            # Remove duplicates while preserving order
            parts = list(dict.fromkeys(parts))

            if len(parts) == 1:
                new_columns.append(parts[0])
            else:
                new_columns.append("_".join(parts))

        df.columns = new_columns

    # Clean normal columns
    else:
        df.columns = [
            str(col).strip()
            for col in df.columns
        ]

    return df
```

```python
def merge_team_table(master_df, new_df, table_name):
    """
    Merge a team-level FBref table into the master dataset.
    """
    
    # Remove duplicate Squad rows if any exist
    new_df = new_df.drop_duplicates(subset="Squad")

    # Don't bring across duplicate columns
    duplicate_columns = [
        col for col in new_df.columns
        if col in master_df.columns and col != "Squad"
    ]
    
    new_df = new_df.drop(columns=duplicate_columns)

    # Merge
    master_df = master_df.merge(
        new_df,
        on="Squad",
        how="left",
        validate="one_to_one"
    )

    print(
        f"{table_name:<55} "
        f"Shape: {master_df.shape}"
    )

    return master_df
```

```python
# ============================================================
# BUILD PRE-MATCH TEAM STATE
# ============================================================

# Make a copy so we don't accidentally alter our clean dataset
fixture_ml = fixtures.copy()

# Store each team's statistics as the season progresses
team_stats = {}

# Columns we will create
state_columns = [
    "Home_Points_Before",
    "Away_Points_Before",
    "Home_GF_Before",
    "Away_GF_Before",
    "Home_GA_Before",
    "Away_GA_Before",
    "Home_Games_Before",
    "Away_Games_Before"
]

for col in state_columns:
    fixture_ml[col] = 0


# ------------------------------------------------------------
# Process fixtures chronologically
# ------------------------------------------------------------

for index, row in fixture_ml.iterrows():

    home = row["Home"]
    away = row["Away"]

    # Create team records if they don't exist yet
    if home not in team_stats:
        team_stats[home] = {
            "points": 0,
            "gf": 0,
            "ga": 0,
            "games": 0
        }

    if away not in team_stats:
        team_stats[away] = {
            "points": 0,
            "gf": 0,
            "ga": 0,
            "games": 0
        }


    # --------------------------------------------------------
    # RECORD STATE BEFORE THE MATCH
    # --------------------------------------------------------

    fixture_ml.at[index, "Home_Points_Before"] = team_stats[home]["points"]
    fixture_ml.at[index, "Away_Points_Before"] = team_stats[away]["points"]

    fixture_ml.at[index, "Home_GF_Before"] = team_stats[home]["gf"]
    fixture_ml.at[index, "Away_GF_Before"] = team_stats[away]["gf"]

    fixture_ml.at[index, "Home_GA_Before"] = team_stats[home]["ga"]
    fixture_ml.at[index, "Away_GA_Before"] = team_stats[away]["ga"]

    fixture_ml.at[index, "Home_Games_Before"] = team_stats[home]["games"]
    fixture_ml.at[index, "Away_Games_Before"] = team_stats[away]["games"]


    # --------------------------------------------------------
    # UPDATE TEAM STATISTICS AFTER THE MATCH
    # --------------------------------------------------------

    home_goals = row["HomeGoals"]
    away_goals = row["AwayGoals"]

    # Goals
    team_stats[home]["gf"] += home_goals
    team_stats[home]["ga"] += away_goals

    team_stats[away]["gf"] += away_goals
    team_stats[away]["ga"] += home_goals

    # Games played
    team_stats[home]["games"] += 1
    team_stats[away]["games"] += 1

    # Points
    if home_goals > away_goals:

        # Home win
        team_stats[home]["points"] += 3

    elif home_goals < away_goals:

        # Away win
        team_stats[away]["points"] += 3

    else:

        # Draw
        team_stats[home]["points"] += 1
        team_stats[away]["points"] += 1
```

```python
# ============================================================
# BUILD LAST-5 FORM
# ============================================================

from collections import defaultdict, deque

# Store previous match results for every team
team_form = defaultdict(lambda: deque(maxlen=5))

# Create columns
form_columns = [
    "Home_Points_Last5",
    "Away_Points_Last5",
    "Home_Wins_Last5",
    "Away_Wins_Last5",
    "Home_Draws_Last5",
    "Away_Draws_Last5",
    "Home_Losses_Last5",
    "Away_Losses_Last5",
    "Home_GF_Last5",
    "Away_GF_Last5",
    "Home_GA_Last5",
    "Away_GA_Last5"
]

for col in form_columns:
    fixture_ml[col] = 0


# ------------------------------------------------------------
# Process matches chronologically
# ------------------------------------------------------------

for index, row in fixture_ml.iterrows():

    home = row["Home"]
    away = row["Away"]

    # Get the previous five matches BEFORE this fixture
    home_previous = list(team_form[home])
    away_previous = list(team_form[away])


    # --------------------------------------------------------
    # HOME FORM
    # --------------------------------------------------------

    fixture_ml.at[index, "Home_Points_Last5"] = sum(
        match["points"] for match in home_previous
    )

    fixture_ml.at[index, "Home_Wins_Last5"] = sum(
        match["result"] == "W" for match in home_previous
    )

    fixture_ml.at[index, "Home_Draws_Last5"] = sum(
        match["result"] == "D" for match in home_previous
    )

    fixture_ml.at[index, "Home_Losses_Last5"] = sum(
        match["result"] == "L" for match in home_previous
    )

    fixture_ml.at[index, "Home_GF_Last5"] = sum(
        match["gf"] for match in home_previous
    )

    fixture_ml.at[index, "Home_GA_Last5"] = sum(
        match["ga"] for match in home_previous
    )


    # --------------------------------------------------------
    # AWAY FORM
    # --------------------------------------------------------

    fixture_ml.at[index, "Away_Points_Last5"] = sum(
        match["points"] for match in away_previous
    )

    fixture_ml.at[index, "Away_Wins_Last5"] = sum(
        match["result"] == "W" for match in away_previous
    )

    fixture_ml.at[index, "Away_Draws_Last5"] = sum(
        match["result"] == "D" for match in away_previous
    )

    fixture_ml.at[index, "Away_Losses_Last5"] = sum(
        match["result"] == "L" for match in away_previous
    )

    fixture_ml.at[index, "Away_GF_Last5"] = sum(
        match["gf"] for match in away_previous
    )

    fixture_ml.at[index, "Away_GA_Last5"] = sum(
        match["ga"] for match in away_previous
    )


    # --------------------------------------------------------
    # DETERMINE RESULT FOR EACH TEAM
    # --------------------------------------------------------

    home_goals = row["HomeGoals"]
    away_goals = row["AwayGoals"]

    if home_goals > away_goals:

        home_result = "W"
        away_result = "L"

        home_points = 3
        away_points = 0

    elif home_goals < away_goals:

        home_result = "L"
        away_result = "W"

        home_points = 0
        away_points = 3

    else:

        home_result = "D"
        away_result = "D"

        home_points = 1
        away_points = 1


    # --------------------------------------------------------
    # ADD THIS MATCH TO EACH TEAM'S HISTORY
    # --------------------------------------------------------

    team_form[home].append({
        "result": home_result,
        "points": home_points,
        "gf": home_goals,
        "ga": away_goals
    })

    team_form[away].append({
        "result": away_result,
        "points": away_points,
        "gf": away_goals,
        "ga": home_goals
    })
```

```python
# ============================================================
# BUILD HOME/AWAY SPECIFIC FORM
# ============================================================

from collections import defaultdict, deque

# Reset venue-specific histories
home_history = defaultdict(lambda: deque(maxlen=5))
away_history = defaultdict(lambda: deque(maxlen=5))


# Create columns
venue_columns = [
    "HomeTeam_HomePoints_Last5",
    "AwayTeam_AwayPoints_Last5",
    "HomeTeam_HomeWins_Last5",
    "AwayTeam_AwayWins_Last5",
    "HomeTeam_HomeDraws_Last5",
    "AwayTeam_AwayDraws_Last5",
    "HomeTeam_HomeLosses_Last5",
    "AwayTeam_AwayLosses_Last5",
    "HomeTeam_HomeGF_Last5",
    "AwayTeam_AwayGF_Last5",
    "HomeTeam_HomeGA_Last5",
    "AwayTeam_AwayGA_Last5"
]

for col in venue_columns:
    fixture_ml[col] = 0


# ============================================================
# PROCESS MATCHES CHRONOLOGICALLY
# ============================================================

for index, row in fixture_ml.iterrows():

    home = row["Home"]
    away = row["Away"]

    # --------------------------------------------------------
    # GET PREVIOUS HOME FORM
    # --------------------------------------------------------

    previous_home_matches = list(home_history[home])

    fixture_ml.at[index, "HomeTeam_HomePoints_Last5"] = sum(
        match["points"] for match in previous_home_matches
    )

    fixture_ml.at[index, "HomeTeam_HomeWins_Last5"] = sum(
        match["result"] == "W"
        for match in previous_home_matches
    )

    fixture_ml.at[index, "HomeTeam_HomeDraws_Last5"] = sum(
        match["result"] == "D"
        for match in previous_home_matches
    )

    fixture_ml.at[index, "HomeTeam_HomeLosses_Last5"] = sum(
        match["result"] == "L"
        for match in previous_home_matches
    )

    fixture_ml.at[index, "HomeTeam_HomeGF_Last5"] = sum(
        match["gf"] for match in previous_home_matches
    )

    fixture_ml.at[index, "HomeTeam_HomeGA_Last5"] = sum(
        match["ga"] for match in previous_home_matches
    )


    # --------------------------------------------------------
    # GET PREVIOUS AWAY FORM
    # --------------------------------------------------------

    previous_away_matches = list(away_history[away])

    fixture_ml.at[index, "AwayTeam_AwayPoints_Last5"] = sum(
        match["points"] for match in previous_away_matches
    )

    fixture_ml.at[index, "AwayTeam_AwayWins_Last5"] = sum(
        match["result"] == "W"
        for match in previous_away_matches
    )

    fixture_ml.at[index, "AwayTeam_AwayDraws_Last5"] = sum(
        match["result"] == "D"
        for match in previous_away_matches
    )

    fixture_ml.at[index, "AwayTeam_AwayLosses_Last5"] = sum(
        match["result"] == "L"
        for match in previous_away_matches
    )

    fixture_ml.at[index, "AwayTeam_AwayGF_Last5"] = sum(
        match["gf"] for match in previous_away_matches
    )

    fixture_ml.at[index, "AwayTeam_AwayGA_Last5"] = sum(
        match["ga"] for match in previous_away_matches
    )


    # --------------------------------------------------------
    # DETERMINE CURRENT MATCH RESULT FOR EACH TEAM
    # --------------------------------------------------------

    home_goals = row["HomeGoals"]
    away_goals = row["AwayGoals"]

    if home_goals > away_goals:

        home_result = "W"
        away_result = "L"

        home_points = 3
        away_points = 0

    elif home_goals < away_goals:

        home_result = "L"
        away_result = "W"

        home_points = 0
        away_points = 3

    else:

        home_result = "D"
        away_result = "D"

        home_points = 1
        away_points = 1


    # --------------------------------------------------------
    # ADD CURRENT MATCH TO VENUE HISTORY
    # --------------------------------------------------------

    home_history[home].append({
        "result": home_result,
        "points": home_points,
        "gf": home_goals,
        "ga": away_goals
    })

    away_history[away].append({
        "result": away_result,
        "points": away_points,
        "gf": away_goals,
        "ga": home_goals
    })
```

```python
# ============================================================
# LOAD + CLEAN ALL 5 PREMIER LEAGUE FIXTURE SEASONS
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path


# ============================================================
# 1. FIXTURE FILES
# ============================================================

project_path = Path(r"C:\Users\karab\PremierLeagueML")

fixture_paths = {
    2022: project_path / "data" / "raw" / "Fixtures" / "2021-2022" / "PL Season.xls",
    2023: project_path / "data" / "raw" / "Fixtures" / "2022-2023" / "PL Season.xls",
    2024: project_path / "data" / "raw" / "Fixtures" / "2023-2024" / "PL Season.xls",
    2025: project_path / "data" / "raw" / "Fixtures" / "2024-2025" / "PL Season.xls",
    2026: project_path / "data" / "raw" / "Fixtures" / "2025-2026" / "PL Season.xls"
}


# ============================================================
# 2. LOAD RAW FIXTURE TABLES
# ============================================================

fixture_raw = {}

for season, path in fixture_paths.items():

    print(f"\nLoading {season}...")

    tables = pd.read_html(path)

    print("Tables found:", len(tables))

    fixture_raw[season] = tables[0]

    print("Raw shape:", tables[0].shape)


# ============================================================
# 3. CLEAN EACH SEASON
# ============================================================

fixtures = {}

for season, df in fixture_raw.items():

    df = df.copy()

    # --------------------------------------------------------
    # Keep only rows containing an actual Home and Away team
    # --------------------------------------------------------

    df = df.dropna(subset=["Home", "Away"]).copy()

    # --------------------------------------------------------
    # Remove rows without a completed score
    # --------------------------------------------------------

    df = df.dropna(subset=["Score"]).copy()

    # --------------------------------------------------------
    # Convert score to string
    # --------------------------------------------------------

    df["Score"] = df["Score"].astype(str)

    # --------------------------------------------------------
    # Fix the incorrectly decoded dash from FBref
    # --------------------------------------------------------

    df["Score"] = (
        df["Score"]
        .str.replace("â€“", "-", regex=False)
        .str.replace("–", "-", regex=False)
        .str.replace("—", "-", regex=False)
    )

    # --------------------------------------------------------
    # Extract Home and Away goals
    # --------------------------------------------------------

    score_parts = df["Score"].str.split("-", expand=True)

    df["HomeGoals"] = pd.to_numeric(
        score_parts[0],
        errors="coerce"
    )

    df["AwayGoals"] = pd.to_numeric(
        score_parts[1],
        errors="coerce"
    )

    # --------------------------------------------------------
    # Remove anything where goals could not be extracted
    # --------------------------------------------------------

    df = df.dropna(
        subset=["HomeGoals", "AwayGoals"]
    ).copy()

    # --------------------------------------------------------
    # Convert goals to integers
    # --------------------------------------------------------

    df["HomeGoals"] = df["HomeGoals"].astype(int)
    df["AwayGoals"] = df["AwayGoals"].astype(int)

    # --------------------------------------------------------
    # Create result
    # --------------------------------------------------------

    df["Result"] = np.select(
        [
            df["HomeGoals"] > df["AwayGoals"],
            df["HomeGoals"] < df["AwayGoals"]
        ],
        [
            "H",
            "A"
        ],
        default="D"
    )

    # --------------------------------------------------------
    # Convert date
    # --------------------------------------------------------

    df["Date"] = pd.to_datetime(
        df["Date"],
        errors="coerce"
    )

    # --------------------------------------------------------
    # Sort chronologically
    # --------------------------------------------------------

    df = df.sort_values(
        ["Date", "Wk"],
        na_position="last"
    ).reset_index(drop=True)

    # --------------------------------------------------------
    # Add project season label
    # --------------------------------------------------------

    df["Season"] = season

    fixtures[season] = df


# ============================================================
# 4. CHECK RESULTS
# ============================================================

print("\n" + "=" * 70)
print("CLEAN FIXTURE SUMMARY")
print("=" * 70)

for season, df in fixtures.items():

    print(
        f"\n{season}: "
        f"{len(df)} matches | "
        f"{df['Date'].min().date()} → "
        f"{df['Date'].max().date()}"
    )

    print("Results:")
    print(df["Result"].value_counts().to_dict())


# ============================================================
# 5. COMBINE ALL SEASONS
# ============================================================

fixtures_all = pd.concat(
    fixtures.values(),
    ignore_index=True
)

print("\n" + "=" * 70)
print("ALL SEASONS")
print("=" * 70)

print("Shape:", fixtures_all.shape)

print("\nMatches by season:")
print(fixtures_all.groupby("Season").size())

print("\nColumns:")
print(fixtures_all.columns.tolist())
```

```python
# ============================================================
# STEP 1 — LOAD AND CLEAN ALL 5 PREMIER LEAGUE SEASONS
# ============================================================

import pandas as pd
import numpy as np
import io
from pathlib import Path


# ============================================================
# 1. PROJECT PATH
# ============================================================

project_path = Path(r"C:\Users\karab\PremierLeagueML")


# ============================================================
# 2. EXACT FIXTURE PATHS
# ============================================================

fixture_paths = {
    2022: project_path / "data" / "raw" / "Fixtures" / "2021-2022 PL Season.xls",
    2023: project_path / "data" / "raw" / "Fixtures" / "2022-2023 PL Season.xls",
    2024: project_path / "data" / "raw" / "Fixtures" / "2023-2024 PL Season.xls",
    2025: project_path / "data" / "raw" / "Fixtures" / "2024-2025 PL Season.xls",
    2026: project_path / "data" / "raw" / "Fixtures" / "2025-2026 PL Season.xls"
}


# ============================================================
# 3. FUNCTION TO LOAD FBREF HTML-DISGUISED-AS-XLS
# ============================================================

def load_fixture_file(path):

    # Files are HTML, not real Excel files
    with open(path, encoding="utf-8") as fh:
        html = fh.read()

    # Parse the HTML table
    df = pd.read_html(io.StringIO(html))[0]

    return df


# ============================================================
# 4. LOAD + CLEAN EACH SEASON
# ============================================================

fixtures = {}

print("=" * 70)
print("LOADING PREMIER LEAGUE FIXTURES")
print("=" * 70)

for season, path in fixture_paths.items():

    print(f"\nLoading {season}...")
    print("Path:", path)

    # Check file exists
    if not path.exists():
        raise FileNotFoundError(
            f"\nFixture file not found:\n{path}"
        )

    # Load HTML table
    df = load_fixture_file(path)

    print("Raw shape:", df.shape)

    # --------------------------------------------------------
    # Remove completely empty spacer rows
    # --------------------------------------------------------

    df = df[df["Date"].notna()].copy()

    # --------------------------------------------------------
    # Date
    # --------------------------------------------------------

    df["Date"] = pd.to_datetime(
        df["Date"],
        errors="coerce"
    )

    # --------------------------------------------------------
    # Matchweek
    # --------------------------------------------------------

    df["Wk"] = pd.to_numeric(
        df["Wk"],
        errors="coerce"
    ).astype(int)

    # --------------------------------------------------------
    # Home / Away
    # --------------------------------------------------------

    df["Home"] = df["Home"].astype(str).str.strip()
    df["Away"] = df["Away"].astype(str).str.strip()

    # --------------------------------------------------------
    # Extract goals
    #
    # This deliberately does NOT split on "-"
    # because FBref uses an en-dash.
    # --------------------------------------------------------

    goals = df["Score"].astype(str).str.extract(
        r"(\d+)\D+(\d+)"
    )

    df["HomeGoals"] = pd.to_numeric(
        goals[0],
        errors="coerce"
    )

    df["AwayGoals"] = pd.to_numeric(
        goals[1],
        errors="coerce"
    )

    # --------------------------------------------------------
    # Remove rows where score could not be interpreted
    # --------------------------------------------------------

    df = df.dropna(
        subset=[
            "Date",
            "Home",
            "Away",
            "HomeGoals",
            "AwayGoals"
        ]
    ).copy()

    # --------------------------------------------------------
    # Convert goals to integers
    # --------------------------------------------------------

    df["HomeGoals"] = df["HomeGoals"].astype(int)
    df["AwayGoals"] = df["AwayGoals"].astype(int)

    # --------------------------------------------------------
    # Match result
    # --------------------------------------------------------

    df["Result"] = np.select(
        [
            df["HomeGoals"] > df["AwayGoals"],
            df["HomeGoals"] < df["AwayGoals"]
        ],
        [
            "H",
            "A"
        ],
        default="D"
    )

    # --------------------------------------------------------
    # Season
    # --------------------------------------------------------

    df["Season"] = season

    # --------------------------------------------------------
    # Sort chronologically
    # --------------------------------------------------------

    df = df.sort_values(
        ["Date", "Wk"]
    ).reset_index(drop=True)

    # --------------------------------------------------------
    # Keep useful columns
    # --------------------------------------------------------

    df = df[
        [
            "Season",
            "Wk",
            "Date",
            "Home",
            "Away",
            "Score",
            "HomeGoals",
            "AwayGoals",
            "Result"
        ]
    ]

    fixtures[season] = df

    print("Clean shape:", df.shape)
    print(
        "Date range:",
        df["Date"].min().date(),
        "→",
        df["Date"].max().date()
    )

    print(
        "Matchweeks:",
        df["Wk"].nunique()
    )


# ============================================================
# 5. COMBINE ALL SEASONS
# ============================================================

fixtures_all = pd.concat(
    fixtures.values(),
    ignore_index=True
)


# ============================================================
# 6. SORT EVERYTHING
# ============================================================

fixtures_all = fixtures_all.sort_values(
    ["Season", "Date", "Wk"]
).reset_index(drop=True)


# ============================================================
# 7. FINAL VALIDATION
# ============================================================

print("\n")
print("=" * 70)
print("FINAL FIXTURE DATASET")
print("=" * 70)

print("\nShape:")
print(fixtures_all.shape)

print("\nMatches per season:")
print(
    fixtures_all.groupby("Season").size()
)

print("\nMatchweeks per season:")
print(
    fixtures_all.groupby("Season")["Wk"].nunique()
)

print("\nMissing values:")
print(
    fixtures_all[
        [
            "Season",
            "Wk",
            "Date",
            "Home",
            "Away",
            "HomeGoals",
            "AwayGoals",
            "Result"
        ]
    ].isna().sum()
)

print("\nNumber of teams:")
print(
    len(
        set(fixtures_all["Home"])
        |
        set(fixtures_all["Away"])
    )
)

print("\nFirst 10 matches:")
print(
    fixtures_all.head(10).to_string(index=False)
)

print("\nLast 10 matches:")
print(
    fixtures_all.tail(10).to_string(index=False)
)
```

```python
# ============================================================
# STEP 1 — BUILD PRE-MATCH HISTORICAL FEATURES
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# Make a working copy
# ------------------------------------------------------------

df = fixtures_all.copy()

# Make absolutely sure matches are chronological
df = df.sort_values(
    ["Date", "Season", "Wk"]
).reset_index(drop=True)

print("=" * 70)
print("BUILDING PRE-MATCH FEATURES")
print("=" * 70)


# ============================================================
# 1. INITIALISE TEAM HISTORY
# ============================================================

team_history = {}

feature_rows = []


# ============================================================
# 2. PROCESS MATCHES IN CHRONOLOGICAL ORDER
# ============================================================

for idx, match in df.iterrows():

    home = match["Home"]
    away = match["Away"]

    season = match["Season"]
    date = match["Date"]

    # --------------------------------------------------------
    # Create team history if this is the team's first match
    # --------------------------------------------------------

    if home not in team_history:
        team_history[home] = {
            "points": 0,
            "goals_for": 0,
            "goals_against": 0,
            "games": 0,
            "last5": [],
            "last_date": None
        }

    if away not in team_history:
        team_history[away] = {
            "points": 0,
            "goals_for": 0,
            "goals_against": 0,
            "games": 0,
            "last5": [],
            "last_date": None
        }


    h = team_history[home]
    a = team_history[away]


    # ========================================================
    # HOME PRE-MATCH FEATURES
    # ========================================================

    if h["games"] > 0:

        home_ppg = h["points"] / h["games"]

        home_gf_pg = (
            h["goals_for"] /
            h["games"]
        )

        home_ga_pg = (
            h["goals_against"] /
            h["games"]
        )

    else:

        home_ppg = np.nan
        home_gf_pg = np.nan
        home_ga_pg = np.nan


    # ========================================================
    # AWAY PRE-MATCH FEATURES
    # ========================================================

    if a["games"] > 0:

        away_ppg = a["points"] / a["games"]

        away_gf_pg = (
            a["goals_for"] /
            a["games"]
        )

        away_ga_pg = (
            a["goals_against"] /
            a["games"]
        )

    else:

        away_ppg = np.nan
        away_gf_pg = np.nan
        away_ga_pg = np.nan


    # ========================================================
    # LAST 5 FORM
    # ========================================================

    if len(h["last5"]) > 0:
        home_last5_ppg = np.mean(h["last5"])
    else:
        home_last5_ppg = np.nan


    if len(a["last5"]) > 0:
        away_last5_ppg = np.mean(a["last5"])
    else:
        away_last5_ppg = np.nan


    # ========================================================
    # REST DAYS
    # ========================================================

    if h["last_date"] is not None:

        home_rest = (
            date - h["last_date"]
        ).days

    else:

        home_rest = np.nan


    if a["last_date"] is not None:

        away_rest = (
            date - a["last_date"]
        ).days

    else:

        away_rest = np.nan


    # ========================================================
    # STORE PRE-MATCH FEATURES
    # ========================================================

    feature_rows.append({

        "Season": season,
        "Wk": match["Wk"],
        "Date": date,

        "Home": home,
        "Away": away,

        "HomeGoals": match["HomeGoals"],
        "AwayGoals": match["AwayGoals"],
        "Result": match["Result"],

        # Historical performance
        "Home_PPG_Before": home_ppg,
        "Away_PPG_Before": away_ppg,

        "Home_GF_PerGame_Before": home_gf_pg,
        "Away_GF_PerGame_Before": away_gf_pg,

        "Home_GA_PerGame_Before": home_ga_pg,
        "Away_GA_PerGame_Before": away_ga_pg,

        # Recent form
        "Home_Last5_PPG": home_last5_ppg,
        "Away_Last5_PPG": away_last5_ppg,

        # Rest
        "Home_Rest_Days": home_rest,
        "Away_Rest_Days": away_rest
    })


    # ========================================================
    # UPDATE HISTORY AFTER THE MATCH
    # ========================================================

    home_goals = match["HomeGoals"]
    away_goals = match["AwayGoals"]


    # Goals
    h["goals_for"] += home_goals
    h["goals_against"] += away_goals

    a["goals_for"] += away_goals
    a["goals_against"] += home_goals


    # Points
    if home_goals > away_goals:

        h["points"] += 3
        a["points"] += 0

        home_points = 3
        away_points = 0

    elif home_goals < away_goals:

        h["points"] += 0
        a["points"] += 3

        home_points = 0
        away_points = 3

    else:

        h["points"] += 1
        a["points"] += 1

        home_points = 1
        away_points = 1


    # Games
    h["games"] += 1
    a["games"] += 1


    # Last 5
    h["last5"].append(home_points)
    a["last5"].append(away_points)

    h["last5"] = h["last5"][-5:]
    a["last5"] = a["last5"][-5:]


    # Last match date
    h["last_date"] = date
    a["last_date"] = date


# ============================================================
# 3. CREATE FEATURE DATAFRAME
# ============================================================

fixture_features = pd.DataFrame(feature_rows)


# ============================================================
# 4. CREATE DIFFERENCE FEATURES
# ============================================================

fixture_features["PPG_Difference"] = (
    fixture_features["Home_PPG_Before"]
    -
    fixture_features["Away_PPG_Before"]
)

fixture_features["GF_PerGame_Difference"] = (
    fixture_features["Home_GF_PerGame_Before"]
    -
    fixture_features["Away_GF_PerGame_Before"]
)

fixture_features["GA_PerGame_Difference"] = (
    fixture_features["Home_GA_PerGame_Before"]
    -
    fixture_features["Away_GA_PerGame_Before"]
)

fixture_features["Last5_PPG_Difference"] = (
    fixture_features["Home_Last5_PPG"]
    -
    fixture_features["Away_Last5_PPG"]
)

fixture_features["Rest_Difference"] = (
    fixture_features["Home_Rest_Days"]
    -
    fixture_features["Away_Rest_Days"]
)


# ============================================================
# 5. CHECK RESULT
# ============================================================

print("\n" + "=" * 70)
print("FEATURE DATASET CREATED")
print("=" * 70)

print("\nShape:")
print(fixture_features.shape)

print("\nColumns:")
print(fixture_features.columns.tolist())

print("\nFirst 15 matches:")
print(
    fixture_features.head(15).to_string(index=False)
)

print("\nMissing values:")
print(
    fixture_features.isna().sum()
)
```

```python
# ============================================================
# STEP 3 — CROSS-SEASON TEAM STRENGTH RATINGS
# ============================================================

import numpy as np
import pandas as pd

print("=" * 70)
print("BUILDING CROSS-SEASON TEAM RATINGS")
print("=" * 70)


# ------------------------------------------------------------
# 1. Start with a copy of our existing feature dataset
# ------------------------------------------------------------

fixture_features = fixture_features.copy()

# Make sure matches are chronologically ordered
fixture_features = fixture_features.sort_values(
    ["Season", "Date"]
).reset_index(drop=True)


# ------------------------------------------------------------
# 2. Calculate each team's season PPG
# ------------------------------------------------------------

team_season_stats = {}

for season in sorted(fixture_features["Season"].unique()):

    season_df = fixture_features[
        fixture_features["Season"] == season
    ].copy()

    teams = set(
        season_df["Home"].unique()
    ).union(
        set(season_df["Away"].unique())
    )

    season_stats = {}

    for team in teams:

        home_matches = season_df[
            season_df["Home"] == team
        ]

        away_matches = season_df[
            season_df["Away"] == team
        ]

        # Points earned
        home_points = (
            (home_matches["HomeGoals"] > home_matches["AwayGoals"]) * 3
            + (home_matches["HomeGoals"] == home_matches["AwayGoals"]) * 1
        ).sum()

        away_points = (
            (away_matches["AwayGoals"] > away_matches["HomeGoals"]) * 3
            + (away_matches["AwayGoals"] == away_matches["HomeGoals"]) * 1
        ).sum()

        total_matches = len(home_matches) + len(away_matches)

        if total_matches > 0:
            ppg = (home_points + away_points) / total_matches
        else:
            ppg = np.nan

        season_stats[team] = ppg

    team_season_stats[season] = season_stats


# ------------------------------------------------------------
# 3. Calculate league-average PPG
# ------------------------------------------------------------

all_ppg = []

for season in team_season_stats:

    all_ppg.extend(
        team_season_stats[season].values()
    )

league_mean_ppg = np.nanmean(all_ppg)

print("\nLeague mean PPG:", round(league_mean_ppg, 3))


# ------------------------------------------------------------
# 4. Create carried-over ratings
#
# We shrink previous-season PPG toward the league mean.
#
# 75% previous strength
# 25% regression toward league average
# ------------------------------------------------------------

SHRINKAGE = 0.25

team_start_rating = {}


for season in sorted(fixture_features["Season"].unique()):

    if season == min(fixture_features["Season"].unique()):

        # First season has no previous season available
        team_start_rating[season] = {}

        continue

    previous_season = season - 1

    previous_stats = team_season_stats.get(
        previous_season,
        {}
    )

    current_start = {}

    for team, previous_ppg in previous_stats.items():

        if pd.isna(previous_ppg):
            rating = league_mean_ppg

        else:
            rating = (
                (1 - SHRINKAGE) * previous_ppg
                + SHRINKAGE * league_mean_ppg
            )

        current_start[team] = rating

    team_start_rating[season] = current_start


# ------------------------------------------------------------
# 5. Add starting ratings to every fixture
# ------------------------------------------------------------

fixture_features["Home_Carryover_Rating"] = np.nan
fixture_features["Away_Carryover_Rating"] = np.nan


for idx, row in fixture_features.iterrows():

    season = row["Season"]

    home = row["Home"]
    away = row["Away"]

    ratings = team_start_rating.get(
        season,
        {}
    )

    fixture_features.at[
        idx,
        "Home_Carryover_Rating"
    ] = ratings.get(
        home,
        league_mean_ppg
    )

    fixture_features.at[
        idx,
        "Away_Carryover_Rating"
    ] = ratings.get(
        away,
        league_mean_ppg
    )


# ------------------------------------------------------------
# 6. Difference between starting strengths
# ------------------------------------------------------------

fixture_features["Carryover_Rating_Difference"] = (
    fixture_features["Home_Carryover_Rating"]
    -
    fixture_features["Away_Carryover_Rating"]
)


# ------------------------------------------------------------
# 7. Check results
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CROSS-SEASON RATINGS CREATED")
print("=" * 70)

print(
    "\nNew columns:"
)

print(
    [
        "Home_Carryover_Rating",
        "Away_Carryover_Rating",
        "Carryover_Rating_Difference"
    ]
)


# ------------------------------------------------------------
# 8. Show GW1 of each season
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("GAMEWEEK 1 CHECK")
print("=" * 70)

gw1 = fixture_features[
    fixture_features["Wk"] == 1
][
    [
        "Season",
        "Home",
        "Away",
        "Home_Carryover_Rating",
        "Away_Carryover_Rating",
        "Carryover_Rating_Difference"
    ]
]

print(
    gw1.to_string(index=False)
)


# ------------------------------------------------------------
# 9. Missing values
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("MISSING VALUES")
print("=" * 70)

print(
    fixture_features[
        [
            "Home_Carryover_Rating",
            "Away_Carryover_Rating",
            "Carryover_Rating_Difference"
        ]
    ].isna().sum()
)


# ------------------------------------------------------------
# 10. Final shape
# ------------------------------------------------------------

print("\nFinal feature dataset shape:")
print(fixture_features.shape)
```
