```python
from pathlib import Path

project_root = Path("..")

overall_file = (
    project_root
    / "data"
    / "raw"
    / "2025 PL Season"
    / "Premier League Overall.xls"
)

print(overall_file)
print(overall_file.exists())
```

```python
import pandas as pd

tables = pd.read_html(overall_file)

print(f"Tables found: {len(tables)}")

overall_df = tables[0]

overall_df.head()
```

```python
overall_df.shape
```

```python
overall_df.columns
```

```python
overall_df.info()
```

```python
overall_df.describe()
```

```python
import matplotlib.pyplot as plt

overall_df.plot.bar(
    x="Squad",
    y="GF",
    figsize=(12,6)
)

plt.xticks(rotation=45)
plt.show()
```

```python
overall_df[["Squad", "GF"]].sort_values("GF", ascending=False)
```

```python
overall_df[["Squad","GF"]].sort_values("GF", ascending=False).head()
```

```python
overall_df[["Squad","GA"]].sort_values("GA")
```

```python
overall_df[["GF","Pts"]].corr()
```

```python
import matplotlib.pyplot as plt

overall_df.plot(
    x="Squad",
    y="GF",
    kind="bar",
    figsize=(12,6)
)

plt.title("Goals Scored by Each Team")
plt.ylabel("Goals")
plt.xticks(rotation=45)

plt.show()
```

```python
overall_df.plot(
    x="Squad",
    y=["GF","GA"],
    kind="bar",
    figsize=(14,6)
)

plt.title("Goals Scored vs Goals Conceded")
plt.xticks(rotation=45)

plt.show()
```
